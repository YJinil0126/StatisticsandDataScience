## Variance

The variance of a random variable $Y$ is defined as:

$$
\mathrm{Var}(Y) = \mathbb{E}[(Y - \mathbb{E}[Y])^2]
$$

It measures the expected squared deviation from the mean.


##### Explicit Calculation of Variance

Given the dataset:

$$
Y = [8,\ 12,\ 10,\ 18,\ 22,\ 20]
$$

We want to compute the **population variance** of $Y$ step by step.

---

 🔹 Step 1: Compute the Mean

The mean (expected value) of $Y$,  $\mathbb{E}[Y]$ is:

$$
\mathbb{E}[Y] = \frac{8 + 12 + 10 + 18 + 22 + 20}{6} = \frac{90}{6} = 15
$$

---
🔹 Step 2: Compute Squared Deviations

We compute each squared deviation from the mean:

$$
\begin{aligned}
(8 - 15)^2 &= 49 \\
(12 - 15)^2 &= 9 \\
(10 - 15)^2 &= 25 \\
(18 - 15)^2 &= 9 \\
(22 - 15)^2 &= 49 \\
(20 - 15)^2 &= 25 \\
\end{aligned}
$$

---

🔹 Step 3: Compute the Variance

Now sum the squared deviations and divide by the number of observations $n = 6$ (since this is **population** variance):

$$
\mathrm{Var}(Y) = \frac{49 + 9 + 25 + 9 + 49 + 25}{6} = \frac{166}{6} \approx 27.67
$$

---

 **Final Summary**

- Mean of $Y$:  
  $$ \bar{Y} = 15 $$

- Population Variance of $Y$:  
  $$ \mathrm{Var}(Y) = 27.67 $$


#### Example

```python
import numpy as np

# Sample data
Y = np.array([8, 12, 10, 18, 22, 20])

# Number of samples
n = len(Y)

# Step 1: Calculate the mean (expected value)
mean_Y = np.mean(Y)

# Step 2: Calculate the squared deviations from the mean
squared_devs = (Y - mean_Y) ** 2

# Step 3: Calculate the variance as the average squared deviation
variance_Y = np.sum(squared_devs) / n  # population variance

print(f"Mean of Y: {mean_Y}")
print(f"Variance of Y: {variance_Y}")
```

## Conditional Expectation and Conditional Variance

The **conditional expectation** of a random variable $Y$ given another random variable $X$ is defined as:

$$
\mathbb{E}[Y \mid X] = g(X)
$$

where $g(X)$ is a function of $X$ representing the expected value of $Y$ when $X$ is known or fixed.

Formally, for discrete $X$:

$$
\mathbb{E}[Y \mid X = x] = \sum_y y \, P(Y = y \mid X = x)
$$

and for continuous $X$:

$$
\mathbb{E}[Y \mid X = x] = \int y \, f_{Y|X}(y \mid x) \, dy,
$$

where $f_{Y|X}(y \mid x)$ is the conditional probability density function of $Y$ given $X = x$.

---

The **conditional variance** of $Y$ given $X$ is:

$$
\mathrm{Var}(Y \mid X) = \mathbb{E} \big[ (Y - \mathbb{E}[Y \mid X])^2 \mid X \big].
$$

This captures the variability of $Y$ within groups defined by $X$.



#### Conditional Mean and Variance – Explicit Calculation

We are given the following dataset:

| Observation | X | Y  |
|-------------|---|----|
| 1           | A | 8  |
| 2           | A | 12 |
| 3           | A | 10 |
| 4           | B | 18 |
| 5           | B | 22 |
| 6           | B | 20 |

---

##### 🔹 Step 1: Conditional Mean ($\mathbb{E}[Y \mid X]$)

**Group A:**

$$
\mathbb{E}[Y \mid X = A] = \frac{8 + 12 + 10}{3} = \frac{30}{3} = 10
$$

**Group B:**

$$
\mathbb{E}[Y \mid X = B] = \frac{18 + 22 + 20}{3} = \frac{60}{3} = 20
$$

---

##### 🔹 Step 2: Conditional Variance ($\mathrm{Var}(Y \mid X)$)

**Group A:**

\
\begin{align*}
\mathrm{Var}(Y \mid X = A) &= \frac{(8 - 10)^2 + (12 - 10)^2 + (10 - 10)^2}{3} \\
&= \frac{4 + 4 + 0}{3} = \frac{8}{3} \approx 2.67
\end{align*}


**Group B:**

\
\begin{align*}
\mathrm{Var}(Y \mid X = B) &= \frac{(18 - 20)^2 + (22 - 20)^2 + (20 - 20)^2}{3} \\
&= \frac{4 + 4 + 0}{3} = \frac{8}{3} \approx 2.67
\end{align*}


---

##### ✅ Final Summary Table

| Group $X$ | $Y$ Values          | $\mathbb{E}[Y \mid X]$ | $\mathrm{Var}(Y \mid X)$ |
|-----------|---------------------|-------------------------|---------------------------|
| A         | 8, 12, 10       | $10$                    | $2.67$                    |
| B         | 18, 22, 20      | $20$                    | $2.67$                    |



## Law of Total Variance

The total variance of a random variable $Y$ can be decomposed into:

$$
\mathrm{Var}(Y) = \underbrace{\mathrm{Var}(\mathbb{E}[Y \mid X])}_{\text{Explained Variance}} + \underbrace{\mathbb{E}[\mathrm{Var}(Y \mid X)]}_{\text{Unexplained Variance}}
$$

---

### 🔹 Explained Variance

This is the variance **explained by** the predictor $X$, i.e., how much the conditional expectation $\mathbb{E}[Y \mid X]$ varies:

$$
\mathrm{Var}(\mathbb{E}[Y \mid X]) = \mathbb{E}[\mathbb{E}[Y \mid X]^2] - \left( \mathbb{E}[\mathbb{E}[Y \mid X]] \right)^2
$$

Note: Since $\mathbb{E}[\mathbb{E}[Y \mid X]] = \mathbb{E}[Y]$, this (by the **law of iterated expectation**) simplifies to:

$$
\mathrm{Var}(\mathbb{E}[Y \mid X]) = \mathbb{E}[\mathbb{E}[Y \mid X]^2] - (\mathbb{E}[Y])^2
$$

---

### 🔹 Unexplained Variance

This is the average **residual variance**, i.e., the variance that remains **within** each group defined by $X$:

$$
\mathbb{E}[\mathrm{Var}(Y \mid X)] = \sum_x P(X = x) \cdot \mathrm{Var}(Y \mid X = x)
$$

In the continuous case:

$$
\mathbb{E}[\mathrm{Var}(Y \mid X)] = \int \mathrm{Var}(Y \mid X = x) \cdot f_X(x) \, dx
$$

---

### ✅ Summary

- $\mathrm{Var}(\mathbb{E}[Y \mid X])$ measures how well $X$ explains variation in $Y$
- $\mathbb{E}[\mathrm{Var}(Y \mid X)]$ measures how much randomness in $Y$ remains after conditioning on $X$
- Together, they add up to the total variance:

$$
\mathrm{Var}(Y) = \mathrm{Var}(\mathbb{E}[Y \mid X]) + \mathbb{E}[\mathrm{Var}(Y \mid X)]
$$



## Student Classroom Example

A classroom of 6 students are examined. Y=Test Scores, and Study Time =X.  We examine how the variance decomposition works and interpretations that can be made from the results.

image.png

#### Variance Decomposition of the classroom

We have student test scores grouped by study time:

| Study Time ($X$) | Test Scores ($Y$)     |
|------------------|----------------------|
| 1                | 75, 77               |
| 2                | 80, 85               |
| 5                | 95, 98               |

Each test score is represented using exact fractions.

---

Step 1: Conditional Mean and Variance

For a random variable $Y$ given $X = x$:

- **Conditional Mean:**

$$
\mathbb{E}[Y \mid X = x] = \frac{1}{n_x} \sum_{i=1}^{n_x} y_i
$$

- **Conditional Variance:**

$$
\mathrm{Var}(Y \mid X = x) = \frac{1}{n_x} \sum_{i=1}^{n_x} (y_i - \mathbb{E}[Y \mid X = x])^2
$$

where $n_x$ is the number of observations in group $x$.

---

Step 2: Group-wise Statistics

| $X$ | $\mathbb{E}[Y \mid X]$ | $\mathrm{Var}(Y \mid X)$ | Count $n_x$ | $\mathbb{P}(X) = \frac{n_x}{N}$ |
|-----|-------------------------|--------------------------|-------------|---------------------------------|
| 1   | $\frac{75 + 77}{2} = 76$       | $\frac{(75 - 76)^2 + (77 - 76)^2}{2} = \frac{1 + 1}{2} = 1$         | 2           | $\frac{2}{6}$                    |
| 2   | $\frac{80 + 85}{2} = 82.5$     | $\frac{(80 - 82.5)^2 + (85 - 82.5)^2}{2} = \frac{6.25 + 6.25}{2} = 6.25$ | 2           | $\frac{2}{6}$                    |
| 5   | $\frac{95 + 98}{2} = 96.5$     | $\frac{(95 - 96.5)^2 + (98 - 96.5)^2}{2} = \frac{2.25 + 2.25}{2} = 2.25$ | 2           | $\frac{2}{6}$                    |

---

Step 3: Overall Mean

$$
\mathbb{E}[Y] = \frac{75 + 77 + 80 + 85 + 95 + 98}{6} = \frac{510}{6} = 85
$$

---

Step 4: Variance Decomposition

**Explained Variance (Between Groups)**

$$
\sum_x \mathbb{P}(X = x) \left( \mathbb{E}[Y \mid X = x] - \mathbb{E}[Y] \right)^2
= \frac{2}{6} (76 - 85)^2 + \frac{2}{6} (82.5 - 85)^2 + \frac{2}{6} (96.5 - 85)^2
$$

Calculating each term:

$$
= \frac{2}{6} \times 81 + \frac{2}{6} \times 6.25 + \frac{2}{6} \times 132.25
= \frac{162 + 12.5 + 264.5}{6} = \frac{439}{6} \approx 73.17
$$

**Unexplained Variance (Within Groups)**

$$
\sum_x \mathbb{P}(X = x) \mathrm{Var}(Y \mid X = x)
= \frac{2}{6} \times 1 + \frac{2}{6} \times 6.25 + \frac{2}{6} \times 2.25
= \frac{2 + 12.5 + 4.5}{6} = \frac{19}{6} \approx 3.17
$$

---

Step 5: Total Decomposed Variance

$$
\mathrm{Var}(Y) = \text{Explained} + \text{Unexplained} = 73.17 + 3.17 = 76.34
$$

---

Step 6: Actual Total Variance

$$
\mathrm{Var}(Y) = \frac{1}{6} \sum_{i=1}^6 (y_i - \mathbb{E}[Y])^2
= \frac{1}{6} \left[(75 - 85)^2 + (77 - 85)^2 + (80 - 85)^2 + (85 - 85)^2 + (95 - 85)^2 + (98 - 85)^2\right]
$$

$$
= \frac{1}{6} \left[100 + 64 + 25 + 0 + 100 + 169 \right] = \frac{458}{6} \approx 76.33
$$

---

 Conclusion

- Explained Variance $\approx 73.17$
- Unexplained Variance $\approx 3.17$
- Total Variance Decomposed $\approx 76.34$
- Actual Total Variance $\approx 76.33$

Minor rounding differences appear but overall the decomposition holds perfectly.

---



```python
from fractions import Fraction
import pandas as pd

# Convert test scores and group by study time using fractions
data_frac = pd.DataFrame({
    'Student': [1, 2, 3, 4, 5, 6],
    'StudyTime': [1, 1, 2, 2, 5, 5],
    'TestScore': list(map(Fraction, [75, 77, 80, 85, 95, 98]))
})

data_frac
```

```python
# Group by StudyTime
grouped_frac = data_frac.groupby('StudyTime')['TestScore']

# Calculate conditional means and variances using fractions
def mean_frac(series):
    return sum(series) / len(series)

def var_frac(series):
    mean = mean_frac(series)
    return sum((x - mean) ** 2 for x in series) / len(series)

stats_frac = grouped_frac.agg([mean_frac, var_frac])
stats_frac.columns = ['E(Y|X)', 'Var(Y|X)']
stats_frac['Count'] = grouped_frac.count()
total_count = stats_frac['Count'].sum()
stats_frac['P(X)'] = stats_frac['Count'].apply(lambda x: Fraction(x, total_count))

stats_frac
```

```python
# Overall mean
overall_mean_frac = mean_frac(data_frac['TestScore'])

# Explained Variance
explained_variance_frac = sum(stats_frac['P(X)'] * (stats_frac['E(Y|X)'] - overall_mean_frac) ** 2)
```

```python
explained_variance_frac
```

```python
# Unexplained Variance
unexplained_variance_frac = sum(stats_frac['P(X)'] * stats_frac['Var(Y|X)'])
```

```python
unexplained_variance_frac
```

```python
# Total decomposed variance
total_decomposed_variance_frac = explained_variance_frac + unexplained_variance_frac

# Actual total variance
actual_total_variance_frac = var_frac(data_frac['TestScore'])

explained_variance_frac, unexplained_variance_frac, total_decomposed_variance_frac, actual_total_variance_frac
```
