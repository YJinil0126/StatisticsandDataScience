#### Probability Basics

1. **Probability Definition**

In probability theory, probability is a measure of the likelihood that a given event will occur.

- **Sample space**: The set of all possible outcomes of a random experiment.
  
  Example: For a fair coin toss, the sample space is:  
  $ \Omega = \{ \text{Heads}, \text{Tails} \} $

- **Event**: A subset of the sample space, i.e., a specific outcome or combination of outcomes.
  
  Example: The event of getting "Heads" is $ E = \{ \text{Heads} \} $.

2. **Basic Probability Axioms**

The following axioms form the foundation of probability:

1. The probability of any event $ A $ is between 0 and 1:
   $ 0 \leq P(A) \leq 1 $

2. The probability of the entire sample space is 1:
   $ P(\Omega) = 1 $

3. For any two mutually exclusive events $ A $ and $ B $:
   $ P(A \cup B) = P(A) + P(B) $

3. **Types of Events**

- **Simple event**: An event that consists of a single outcome.
- **Compound event**: An event that consists of more than one outcome.
- **Mutually exclusive events**: Two events that cannot both occur at the same time.
  Example: In a coin toss, "Heads" and "Tails" are mutually exclusive.
- **Independent events**: Two events are independent if the occurrence of one does not affect the probability of the other.
  
4. **Computing Probability**

For **discrete random variables**, the probability of an event $ A $ is given by:

$ P(A) = \frac{\text{Number of favorable outcomes for A}}{\text{Total number of possible outcomes}} $

Example: The probability of drawing an Ace from a standard deck of cards is:

$ P(\text{Ace}) = \frac{4}{52} = \frac{1}{13} $

5. **Complementary Events**

The complement of an event $ A $ is the event that $ A $ does not occur, denoted as $ A^c $.

- The probability of the complement of $ A $ is:

$ P(A^c) = 1 - P(A) $

Example: If the probability of raining tomorrow is 0.3, the probability that it won’t rain is:

$ P(\text{Not rain}) = 1 - 0.3 = 0.7 $

6. **Example: Probability of Rolling a Die**

Consider a fair six-sided die. The sample space is:

$ \Omega = \{1, 2, 3, 4, 5, 6\} $

The probability of rolling a 4:

$ P(\text{4}) = \frac{1}{6} $

The probability of rolling an even number:

$ P(\text{Even}) = \frac{3}{6} = \frac{1}{2} $

#### Linearity of Expectations

Two_red_dice_01.svg.png

 **Linearity of Expectation**

The Linearity of Expectation states that the expected value of a sum of random variables is equal to the sum of their expected values. This property holds **regardless of whether the random variables are independent**.

Let $X_1, X_2, \dots, X_n$ be random variables. Then:

$$
E[X_1 + X_2 + \dots + X_n] = E[X_1] + E[X_2] + \dots + E[X_n]
$$

If constants $c_1, c_2, \dots, c_n$ are multiplied with the variables, then:

$$
E[c_1 X_1 + c_2 X_2 + \dots + c_n X_n] = c_1 E[X_1] + c_2 E[X_2] + \dots + c_n E[X_n]
$$

---

**Example 1: Linearity of Expectation (Sum of Dice)**

Let $X_1$ and $X_2$ represent the outcomes of rolling two fair six-sided dice.

Each die has expected value:

$$
E[X_1] = E[X_2] = \frac{1 + 2 + 3 + 4 + 5 + 6}{6} = 3.5
$$

Define a new variable:

$$
S = X_1 + X_2
$$

By **linearity of expectation**:

$$
E[S] = E[X_1 + X_2] = E[X_1] + E[X_2] = 3.5 + 3.5 = 7
$$

This demonstrates how the expected value of a **sum** equals the **sum of expected values**.

---

**Example 2: Scaling a Random Variable**

Let $X$ be the outcome of a fair six-sided die. The expected value is:

$$
E[X] = \frac{1 + 2 + 3 + 4 + 5 + 6}{6} = 3.5
$$

Now suppose we define $Y = 5X$. Then:

$$
E[Y] = E[5X] = 5 \cdot E[X] = 5 \cdot 3.5 = 17.5
$$

This demonstrates how constants can be factored out of the expectation.

---




**Example 3: Linearity Holds Even with Dependence**

Suppose you draw two cards from a standard 52-card deck **without replacement**.

Define:

- $X_1 = 1$ if the first card is a heart, else $0$
- $X_2 = 1$ if the second card is a heart, else $0$

These are **dependent** random variables (since there's no replacement), but:

$$
E[X_1 + X_2] = E[X_1] + E[X_2] = \frac{13}{52} + \frac{13}{52} = \frac{26}{52} = 0.5
$$

Linearity of expectation still holds, even though $X_1$ and $X_2$ are not independent.


```python
import random
import matplotlib.pyplot as plt

# Simulate drawing 2 cards without replacement
def simulate_two_card_draw(trials=10000):
    hearts_counts = []

    for _ in range(trials):
        deck = ['heart'] * 13 + ['non-heart'] * 39
        drawn_cards = random.sample(deck, 2)
        count_hearts = drawn_cards.count('heart')
        hearts_counts.append(count_hearts)

    return hearts_counts

# Run simulation
trials = 10000
hearts_counts = simulate_two_card_draw(trials)

# Calculate average number of hearts per trial
average_hearts = sum(hearts_counts) / trials

# Plot histogram
plt.figure(figsize=(8, 5))
plt.hist(hearts_counts, bins=[-0.5, 0.5, 1.5, 2.5], rwidth=0.8, color='skyblue', edgecolor='black')
plt.xticks([0, 1, 2])
plt.xlabel("Number of Hearts in 2 Cards")
plt.ylabel("Frequency")
plt.title(f"Distribution of Hearts in 2-Card Draws ({trials} Trials)\nAverage = {average_hearts:.3f}")
plt.grid(axis='y')
plt.show()
```

#### Variance and Standard Deviation

**Variance and Standard Deviation**

Measures spread around the mean:

$$
\text{Var}(X) = \mathbb{E}[(X - \mathbb{E}[X])^2] = \mathbb{E}[X^2] - (\mathbb{E}[X])^2
$$

$$
\text{SD}(X) = \sqrt{\text{Var}(X)}
$$

```python
import numpy as np

# Simulate a fair 6-sided die roll
n = 10000  # Number of rolls
outcomes = np.random.randint(1, 7, size=n)

# Compute the expected value (mean)
expected_value = np.mean(outcomes)

# Compute the expected value of X^2 (second moment)
expected_value_squared = np.mean(outcomes**2)

# Compute the variance using Var(X) = E[X^2] - (E[X])^2
variance = expected_value_squared - expected_value**2

print("Expected value (E[X]):", expected_value)
print("Expected value of X^2 (E[X^2]):", expected_value_squared)
print("Variance (Var[X]):", variance)
```

**Joint Probability**

The joint probability of two events $A$ and $B$, denoted as $P(A \cap B)$ or $P(A, B)$, is the probability that both events occur at the same time.

Mathematically, it is defined as:

$P(A \cap B) = \text{Probability that both } A \text{ and } B \text{ occur}$


#### Conditional Probability

**Conditional Probability**

The probability of $A$ given $B$ is:

$$
P(A \mid B) = \frac{P(A \cap B)}{P(B)}
$$

Used in risk modeling, Bayesian inference, etc.

##### Dice Example (Discrete)

We want to compute the conditional probability:

$$P(X = 5 \mid X > 3)$$

This is the probability that the die shows a 5 given that it is greater than 3.

The sample space of a fair six-sided die is:

$$\{1, 2, 3, 4, 5, 6\}$$

Conditioning on $X > 3$, the new sample space becomes:

$$\{4, 5, 6\}$$

There is only one favorable outcome (5) among the 3 possible outcomes.

So:

$$P(X = 5 \mid X > 3) = \frac{1}{3}$$


```python
# All outcomes of a fair 6-sided die
outcomes = [1, 2, 3, 4, 5, 6]

# Condition: X > 3
conditioned_outcomes = [x for x in outcomes if x > 3]

# Total number of favorable outcomes under the condition
favorable = conditioned_outcomes.count(5)

# Total number of possible outcomes under the condition
total = len(conditioned_outcomes)

# Conditional probability
conditional_prob = favorable / total

print("P(X = 5 | X > 3) =", conditional_prob)
```

##### Another Example

We want to compute the conditional probability distribution of $X$ given that $X > 4$:

$$P(X \mid X > 4)$$

The sample space of a fair six-sided die is:

$$\{1, 2, 3, 4, 5, 6\}$$

Conditioning on $X > 4$, the new sample space becomes:

$$\{5, 6\}$$

Since the die is fair, each outcome in this conditional space is equally likely. Therefore, the conditional distribution is:

$$
P(X = 5 \mid X > 4) = \frac{1}{2}, \quad
P(X = 6 \mid X > 4) = \frac{1}{2}
$$

Or more compactly:

$$
P(X \mid X > 4) =
\begin{cases}
\frac{1}{2} & \text{if } X = 5 \text{ or } X = 6 \\
0 & \text{otherwise}
\end{cases}
$$


##### Normal Distribution (Continuous)

**Key Insight**  
In a continuous distribution, the probability of any exact value is zero:  
$P(X = x) = 0$  

So instead of conditioning on a single value, we consider small neighborhoods:  
$P(Y \mid X \in (x - \varepsilon, x + \varepsilon))$  

This makes conditional probability meaningful, and as $\varepsilon \to 0$, the conditional probability approaches the conditional density.
  
---

**Example with the Normal Distribution**  

Suppose $X \sim \mathcal{N}(\mu, \sigma^2)$, and we are interested in the conditional probability $P(Y \mid X = x)$.  

Since $P(X = x) = 0$, this is not well-defined.  

Instead, we approximate it using an $\varepsilon$-neighborhood:  
$P(Y \mid X \in (x - \varepsilon, x + \varepsilon))$  

This captures the idea that in continuous distributions, we must think in terms of intervals (neighborhoods), not point values.  

In probability theory, the conditional density is defined as:  
$$f_{Y \mid X}(y \mid x) = \dfrac{f_{X,Y}(x, y)}{f_X(x)}$$  

This definition reflects the limit behavior as the neighborhood around $x$ shrinks to zero.


Conditional Distribution $P(X | Y = 5)$

We are interested in estimating the conditional distribution of $X$ given $Y = 5$. In the context of continuous random variables, this is represented as:

$$ P(X | Y = 5) $$

Given the relationship between $X$ and $Y$ in the simulation:

$$ X = 3Y + \epsilon $$

where $\epsilon \sim \mathcal{N}(0, 1)$ is the noise term, we can infer that $X$ is linearly dependent on $Y$. Specifically, when $Y$ is approximately 5, the value of $X$ is centered around:

$$ X = 3 \times 5 + \epsilon = 15 + \epsilon $$

Thus, the conditional distribution $P(X | Y = 5)$ is approximately normal with a mean of 15 and a standard deviation equal to the standard deviation of the noise term, $\epsilon$, which is 1. Therefore:

$$ P(X | Y = 5) \sim \mathcal{N}(15, 1) $$

In the code, we estimate this distribution by selecting values of $X$ corresponding to $Y$ values within a small window around 5 (i.e., $5 \pm \epsilon$), which gives us the empirical distribution of $X$ when $Y \approx 5$.


```python
import numpy as np
import matplotlib.pyplot as plt

# Simulate joint data for X and Y
np.random.seed(42)
n = 100000
Y = np.random.normal(5, 1, n)  # Y ~ N(5, 1)
X = 3 * Y + np.random.normal(0, 1, n)  # X depends on Y: X = 3Y + noise

# Estimate P(X | Y = 5)
# Since Y is continuous, we'll use a small window around 5
epsilon = 0.05
mask = (Y > 5 - epsilon) & (Y < 5 + epsilon)
X_given_Y5 = X[mask]

# Visualize the conditional distribution P(X | Y = 5)
plt.hist(X_given_Y5, bins=50, density=True, alpha=0.7, color='skyblue')
plt.title("Estimated $P(X | Y = 5)$")
plt.xlabel("X")
plt.ylabel("Density")
plt.grid(True)
plt.show()
```

#### Conditional Expectation

**Conditional Expectation**

The expected value of $X$ given $Y$ is:

$$
\mathbb{E}[X \mid Y] = \int x \cdot f_{X \mid Y}(x \mid y) \, dx
$$

This forms the basis of regression and many filtering techniques in finance

##### Dice Example

```python
# Possible die outcomes
outcomes = [1, 2, 3, 4, 5, 6]

# Condition: X > 3
conditioned_outcomes = [x for x in outcomes if x > 3]

# All outcomes are equally likely, so use uniform probability
prob = 1 / len(conditioned_outcomes)

# Compute conditional expectation
conditional_expectation = sum(x * prob for x in conditioned_outcomes)

print("E[X | X > 3] =", conditional_expectation)
```

Predicting House Prices using Linear Regression

In this example, we will predict house prices based on the size of the house using a simple linear regression model.

Assume the linear relationship between the house price $Y$ and the house size $X$ is given by:

$$
Y = \beta_0 + \beta_1 X + \epsilon
$$

Where:
- $Y$ is the house price.
- $X$ is the house size (in square feet).
- $\beta_0$ is the intercept (the price when the house size $X = 0$).
- $\beta_1$ is the slope (the change in price per square foot).
- $\epsilon$ is the error term (random noise or other unaccounted factors).

Let’s assume the following values:
- $\beta_0 = 50{,}000$
- $\beta_1 = 200$
- $\epsilon \sim \mathcal{N}(0, \sigma^2)$ (mean-zero noise)

The conditional expectation $E[Y \mid X = x]$ is the expected house price given a specific size $x$. In a linear regression model, this is:

$$
E[Y \mid X = x] = \beta_0 + \beta_1 \cdot x
$$

Suppose the house size is $X = 1500$ square feet:

$$
E[Y \mid X = 1500] = 50{,}000 + 200 \cdot 1500 = 50{,}000 + 300{,}000 = 350{,}000
$$

So, the expected price for a 1,500 sq ft house is **$350,000**.

In real-world data, we add randomness using $\epsilon$. For example, if $\epsilon = 10{,}000$, then the actual price could be:

$$
Y = 350{,}000 + 10{,}000 = 360{,}000
$$

This value represents one possible observation around the expected value due to random variation.

##### Full Explanation

The **Law of Total Probability** provides a way to compute the probability of an event $A$ by partitioning the sample space into mutually exclusive events $B_1, B_2, \dots, B_n$ that cover the whole space (i.e., they form a partition).

Formally, if $B_1, B_2, \dots, B_n$ are mutually exclusive and exhaustive events such that:

- $B_i \cap B_j = \emptyset$ for $i \neq j$
- $\bigcup_{i=1}^{n} B_i = \Omega$ (the entire sample space)

then for any event $A$:

$P(A) = \sum_{i=1}^{n} P(A \mid B_i) P(B_i)$

This rule is especially useful when it is difficult to compute $P(A)$ directly, but you know how $A$ behaves under different conditions $B_i$.


image.png

**Example**:

Suppose a factory has three machines $M_1$, $M_2$, and $M_3$ producing widgets. Let:

- $P(M_1) = 0.3$, $P(M_2) = 0.45$, $P(M_3) = 0.25$
- The probability a widget is defective given it was produced by each machine is:
  - $P(D \mid M_1) = 0.01$
  - $P(D \mid M_2) = 0.02$
  - $P(D \mid M_3) = 0.03$

To find the total probability that a randomly selected widget is defective:

$P(D) = P(D \mid M_1) P(M_1) + P(D \mid M_2) P(M_2) + P(D \mid M_3) P(M_3)$

$P(D) = 0.01 \cdot 0.3 + 0.02 \cdot 0.45 + 0.03 \cdot 0.25$

$P(D) = 0.003 + 0.009 + 0.0075 = 0.0195$

So the total probability that a randomly selected widget is defective is $0.0195$.


#### Bayes Theorem

Bayes' Theorem describes the probability of an event based on prior knowledge of conditions related to the event.

The formula is:

$P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}$

Where:

- $P(A|B)$ is the **posterior**: the probability of hypothesis $A$ given the observed evidence $B$.
- $P(B|A)$ is the **likelihood**: the probability of observing the evidence $B$ given that $A$ is true.
- $P(A)$ is the **prior**: the initial probability of hypothesis $A$ before seeing the evidence.
- $P(B)$ is the **marginal likelihood** or **evidence**: the total probability of observing $B$ under all possible hypotheses.


You see that the ground is wet. What is the probability that it rained?

Bayes' Theorem:

$P(R|W) = \frac{P(W|R) \cdot P(R)}{P(W)}$

Where:

- $P(R|W)$ is the posterior: probability it rained given the ground is wet.
- $P(W|R)$ is the likelihood: probability the ground is wet given it rained.
- $P(R)$ is the prior: probability it rains on any given day.
- $P(W)$ is the evidence: total probability that the ground is wet.

We compute $P(W)$ using the law of total probability:

$P(W) = P(W|R) \cdot P(R) + P(W|\lnot R) \cdot P(\lnot R)$


```python
# Given probabilities
P_R = 0.2  # P(R) - Prior: Probability it rained
P_not_R = 1 - P_R  # P(¬R)

P_W_given_R = 0.9  # P(W | R) - Likelihood: wet ground if it rained
P_W_given_not_R = 0.1  # P(W | ¬R) - Likelihood: wet ground if it didn't rain

# Marginal likelihood (evidence)
P_W = P_W_given_R * P_R + P_W_given_not_R * P_not_R

# Posterior: P(R | W) - probability it rained given wet ground
P_R_given_W = (P_W_given_R * P_R) / P_W

print("P(R | W) - Probability it rained given the ground is wet:", P_R_given_W)


```

#### Independence

**Independence**

Two events $A$ and $B$ are independent if:

$$
P(A \cap B) = P(A) \cdot P(B)
$$

Two random variables $X$ and $Y$ are independent if:

$$
f_{X,Y}(x, y) = f_X(x) \cdot f_Y(y)
$$

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

# 1. Independent Events
# Tossing a fair coin twice
coin1 = np.random.choice(['H', 'T'], size=1000)
coin2 = np.random.choice(['H', 'T'], size=1000)

# Event A: coin1 is 'H', Event B: coin2 is 'H'
A = (coin1 == 'H')
B = (coin2 == 'H')

# Estimate P(A), P(B), P(A and B), P(A)*P(B)
P_A = np.mean(A)
P_B = np.mean(B)
P_A_and_B = np.mean(A & B)
P_A_times_P_B = P_A * P_B

# 2. Independent Random Variables
X = np.random.normal(0, 1, 1000)
Y = np.random.normal(5, 2, 1000)  # Independent of X

# 3. Dependent Random Variables
Z = 2 * X + np.random.normal(0, 0.5, 1000)

# Scatter plots to show independence vs dependence
fig, axs = plt.subplots(1, 2, figsize=(12, 4))

axs[0].scatter(X, Y, alpha=0.5)
axs[0].set_title("Independent Random Variables: X vs Y")
axs[0].set_xlabel("X")
axs[0].set_ylabel("Y")

axs[1].scatter(X, Z, alpha=0.5, color='orange')
axs[1].set_title("Dependent Random Variables: X vs Z")
axs[1].set_xlabel("X")
axs[1].set_ylabel("Z")

plt.tight_layout()
plt.show()

# Correlations to confirm
corr_XY = np.corrcoef(X, Y)[0, 1]
corr_XZ = np.corrcoef(X, Z)[0, 1]

(P_A, P_B, P_A_and_B, P_A_times_P_B, corr_XY, corr_XZ)

```

##### Application of Independence in Joint Probability

**Problem:**
What is the probability of rolling a 4 on the second die, given that the first die shows a 3?

In terms of conditional probability, we want to find:
$$
P(\text{Second die is 4} \mid \text{First die is 3})
$$

**Conditional Probability Formula:**
The conditional probability formula is given by:
$$
P(A \mid B) = \frac{P(A \cap B)}{P(B)}
$$
Where:
- $A$ is the event we are interested in (second die is 4),
- $B$ is the condition (first die is 3),
- $P(A \cap B)$ is the joint probability of both events occurring (first die is 3 and second die is 4),
- $P(B)$ is the probability of the condition (first die is 3).

**Solution:**
1. $P(B)$: The probability that the first die shows a 3 is:
$$
P(B) = \frac{1}{6}
$$
2. $P(A \cap B)$: The probability that the first die shows a 3 and the second die shows a 4 is:
$$
P(A \cap B) = P(\text{First die is 3}) \times P(\text{Second die is 4}) = \frac{1}{6} \times \frac{1}{6} = \frac{1}{36}
$$
3. Now we can apply the conditional probability formula:
$$
P(\text{Second die is 4} \mid \text{First die is 3}) = \frac{P(\text{First die is 3 and Second die is 4})}{P(\text{First die is 3})}
$$
$$
P(\text{Second die is 4} \mid \text{First die is 3}) = \frac{\frac{1}{36}}{\frac{1}{6}} = \frac{1}{6}
$$

**Conclusion:**
The conditional probability that the second die shows a 4, given that the first die shows a 3, is $ \frac{1}{6} $.

This makes sense because, regardless of what the first die shows, the second die still has an equal chance of showing any number from 1 to 6, so the probability of rolling a 4 on the second die is $ \frac{1}{6} $.

##### Regression and Covariance (A Probabilistic View)

Linear regression models the relationship between two variables by fitting a straight line to the observed data. The key assumption is that one variable, say $X$, can be used to predict another variable $Y$.

The basic form of a simple linear regression model is:

$Y = \beta_0 + \beta_1 X + \varepsilon$

where:
- $\beta_0$ is the intercept
- $\beta_1$ is the slope (coefficient)
- $\varepsilon$ is the error term

To estimate $\beta_1$, we look at how $X$ and $Y$ vary together, which brings us to **covariance**.

Covariance measures the joint variability of two random variables:

$\mathrm{Cov}(X, Y) = \frac{1}{n} \sum_{i=1}^{n} (X_i - \bar{X})(Y_i - \bar{Y})$

If $\mathrm{Cov}(X, Y)$ is positive, $X$ and $Y$ tend to increase together. If it's negative, as $X$ increases, $Y$ tends to decrease.

The slope $\beta_1$ in simple linear regression can be expressed using covariance and variance:

$\beta_1 = \frac{\mathrm{Cov}(X, Y)}{\mathrm{Var}(X)}$

and:

$\beta_0 = \bar{Y} - \beta_1 \bar{X}$

The variance of $X$ is:

$\mathrm{Var}(X) = \frac{1}{n} \sum_{i=1}^{n} (X_i - \bar{X})^2$

So the linear regression model becomes:

$Y = (\bar{Y} - \beta_1 \bar{X}) + \beta_1 X$

This line minimizes the **sum of squared errors** (SSE) between the predicted and actual values:

$SSE = \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2$

The goodness of fit is commonly measured by the coefficient of determination, $R^2$:

$R^2 = 1 - \frac{\sum (Y_i - \hat{Y}_i)^2}{\sum (Y_i - \bar{Y})^2}$

Where:
- $\hat{Y}_i$ is the predicted value from the model
- $\bar{Y}$ is the mean of $Y$

An $R^2$ close to 1 indicates that the model explains most of the variability in $Y$.

To summarize:
- Covariance helps determine the **direction** of the relationship.
- The slope $\beta_1$ uses covariance normalized by the variance of $X$.
- The intercept aligns the regression line with the means of $X$ and $Y$.
- The line is chosen to minimize the prediction error (SSE).






```python

import numpy as np
import matplotlib.pyplot as plt

# Example data
X = np.array([1, 2, 3, 4, 5])
Y = np.array([2, 4, 5, 4, 5])

# Mean
X_mean = np.mean(X)
Y_mean = np.mean(Y)

# Covariance and Variance
cov_XY = np.mean((X - X_mean) * (Y - Y_mean))
var_X = np.mean((X - X_mean)**2)

# Slope and Intercept
beta_1 = cov_XY / var_X
beta_0 = Y_mean - beta_1 * X_mean

# Predictions
Y_pred = beta_0 + beta_1 * X

# Plot
plt.scatter(X, Y, label='Data')
plt.plot(X, Y_pred, color='red', label='Regression line')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.title('Linear Regression Using Covariance')
plt.show()
```

#### Joint and Marginal Probability (Discrete Case)

```python
import pandas as pd

# Sample data: each row is a website visit
data = [
    {'Device': 'Mobile', 'Action': 'Purchase'},
    {'Device': 'Mobile', 'Action': 'No Purchase'},
    {'Device': 'Desktop', 'Action': 'Purchase'},
    {'Device': 'Mobile', 'Action': 'Purchase'},
    {'Device': 'Desktop', 'Action': 'No Purchase'},
    {'Device': 'Mobile', 'Action': 'No Purchase'},
    {'Device': 'Mobile', 'Action': 'Purchase'},
    {'Device': 'Desktop', 'Action': 'Purchase'},
    {'Device': 'Desktop', 'Action': 'No Purchase'},
    {'Device': 'Mobile', 'Action': 'No Purchase'},
]

df = pd.DataFrame(data)

# Joint frequency table
joint_counts = pd.crosstab(df['Device'], df['Action'])
print("Joint Frequency Table:")
print(joint_counts)

# Convert to joint probability distribution
joint_probs = joint_counts / joint_counts.values.sum()
print("\nJoint Probability Distribution:")
print(joint_probs)

# Marginal distributions
marginal_device = joint_probs.sum(axis=1)  # Sum over actions
marginal_action = joint_probs.sum(axis=0)  # Sum over devices

print("\nMarginal Distribution of Device:")
print(marginal_device)

print("\nMarginal Distribution of Action:")
print(marginal_action)
```

#### Convergence In Distribution

Convergence in Distribution

Let $X_n$ be a sequence of random variables and $X$ another random variable.  
We say that $X_n$ **converges in distribution** to $X$, written as:

$$
X_n \xrightarrow{d} X
$$

This means that the **cumulative distribution function (CDF)** of $X_n$ converges to the CDF of $X$ at every point of continuity of $F_X$. That is,

$$
\lim_{n \to \infty} F_{X_n}(x) = F_X(x)
$$

for all $x$ where $F_X$ is continuous.

---

🔍 Intuition

- You are not saying that $X_n$ becomes equal to $X$.
- You are saying that the **distribution of $X_n$ becomes closer and closer** to the distribution of $X$.
- This is useful when approximating complicated distributions using known ones (e.g., normal distribution via CLT).

--📉 Central Limit Theorem (CLT) Example

Let $\bar{X}_n$ be the sample mean of $n$ i.i.d. variables with mean $\mu$ and variance $\sigma^2$. Then,

$$
\frac{\bar{X}_n - \mu}{\sigma / \sqrt{n}} \xrightarrow{d} \mathcal{N}(0, 1)
$$

This means the standardized sample mean **converges in distribution** to a standard normal distribution as $n \to \infty$.

Even though the sample mean is random, its **distribution shape becomes more and more like a bell curve**.


##### Central Limit Theorem Demonstration

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
# Create a skewed (exponential) population
np.random.seed(42)
population = np.random.exponential(scale=2, size=100000)

# Plot the population distribution
plt.figure(figsize=(10, 4))
sns.histplot(population, bins=50, kde=True, color='skyblue')
plt.title("Skewed Population Distribution (Exponential)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
sample_size = 30
num_samples = 1000

sample_means = []

for _ in range(num_samples):
    sample = np.random.choice(population, size=sample_size, replace=False)
    sample_means.append(np.mean(sample))
```

##### Central Limit Theorem and Its Justification

---

1. Confidence Intervals for the Mean

To construct a **confidence interval** for the population mean $\mu$, we need to estimate the **range of values** within which we believe $\mu$ lies, based on our sample mean $\bar{x}$.

Using the Central Limit Theorem (CLT), we know that the **sampling distribution of the sample mean** $\bar{x}$ will be approximately normal, regardless of the original population's distribution, for sufficiently large $n$. Specifically:

$$
\bar{x} \sim \mathcal{N}\left(\mu, \frac{\sigma^2}{n} \right)
$$

Thus, we can construct a **confidence interval** for the population mean $\mu$ using the normal distribution:

$$
\bar{x} \pm z^* \cdot \frac{\sigma}{\sqrt{n}}
$$

where $z^*$ is the critical value from the standard normal distribution for the desired confidence level (e.g., $z^* = 1.96$ for a 95% confidence interval).

Since the CLT tells us that $\bar{x}$ is approximately normal, we can confidently use this formula to compute confidence intervals, even if the population itself is not normally distributed, **as long as $n$ is large enough**.

---

2. Z-tests and T-tests

In **hypothesis testing**, we want to test statements about the population mean $\mu$. For example, we might test if the true mean is equal to some value $\mu_0$.

This requires us to compute the test statistic, which for the Z-test is given by:

$$
z = \frac{\bar{x} - \mu_0}{\frac{\sigma}{\sqrt{n}}}
$$

For the **t-test**, we replace the population standard deviation $\sigma$ with the sample standard deviation $s$, as follows:

$$
t = \frac{\bar{x} - \mu_0}{\frac{s}{\sqrt{n}}}
$$

Both of these tests assume that the **sampling distribution of the sample mean** is approximately normal, which is guaranteed by the CLT. Even if the population distribution is not normal, the CLT allows us to assume the sampling distribution of the mean will be normal **for large $n$**, so we can use these tests to make inferences about the population mean $\mu$.

---

3. Estimating the Population Mean

The sample mean $\bar{x}$ is an **unbiased estimator** of the population mean $\mu$, meaning that on average, $\bar{x}$ will equal $\mu$. Additionally, the **variability** (or **standard error**) of $\bar{x}$ is known and can be calculated as:

$$
\text{SE} = \frac{\sigma}{\sqrt{n}}
$$

Because of the CLT, we know that as we draw more samples, the distribution of $\bar{x}$ will **converge to the population mean** $\mu$, and we can estimate $\mu$ using $\bar{x}$ with a known degree of uncertainty (the standard error).

Without the CLT, we wouldn't be able to justify using the sample mean to estimate the population mean unless we knew the population was normal. The CLT ensures that **even with a non-normal population**, the sample mean behaves predictably and can be used to make reliable estimates of $\mu$.
