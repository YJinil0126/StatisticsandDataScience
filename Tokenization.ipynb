### Byte Pair Encoding

#### Compress the bytes_list with BYTE PAIR ENCODING

#### Pair that occurrs most frequently, then we mint a new token

```python
text="Ｕｎｉｃｏｄｅ! 🅤🅝🅘🅒🅞🅓🅔‽ 🇺‌🇳‌🇮‌🇨‌🇴‌🇩‌🇪! 😄 The very name strikes fear and awe into the hearts of programmers worldwide. "
tokens=text.encode("UTF-8")
tokens=list(map(int,tokens))
print("---")
print(text)
print("length",len(text))
print("---")
print(tokens)
print("length",len(tokens))
```

#### THe simple asci characters become a single byte more complex characters become more like 4

```python
def get_pairs(ids):
    counts={}
    for pair in zip(ids,ids[1:]):
     counts[pair]=counts.get(pair,0)+1
    return counts

stats=get_pairs(tokens)
print(stats)
```

```python
print(sorted(((v,k) for k,v in stats.items()),reverse=True))
```

```python
chr(240),chr(159)
```

```python
top_pair=max(stats,key=stats.get)
top_pair
```

```python
def merge(ids,pair,idx):
    newids=[]
    i=0
    while i< len(ids):
        if i< len(ids)-1 and ids[i]==pair[0] and ids[i+1]==pair[1]:
            newids.append(idx)
            i+=2 #skip over entire pair
        else:
            newids.append(ids[i])
            i+=1
    return newids

print(merge([5,6,6,7,9,1],(6,7),99)) #replace 6,7 with 99
```
